```python
# Printing all even numbers from 1 to 100
print("Even numbers from 1 to 100:")
for num in range(1, 101):
    if num % 2 == 0:
        print(num, end=" ")
print("\n") # Newline for clarity

# Bonus: Printing odd numbers from 1 to 100
print("Odd numbers from 1 to 100:")
for num in range(1, 101):
    if num % 2 != 0:
        print(num, end=" ")
print()
```

```python
correct_pin = "1234"
attempts = 0
max_attempts = 3

# Loop runs while attempts are within limit
while attempts < max_attempts:
    entered_pin = input("Enter your 4-digit ATM PIN: ")
    attempts += 1

    if entered_pin == correct_pin:
        print("Transaction Allowed")
        break
    else:
        remaining = max_attempts - attempts
        if remaining > 0:
            print(f"Incorrect PIN. You have {remaining} attempts left.")
        else:
            print("Card Blocked")
```
